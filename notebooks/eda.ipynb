# EDA: Product Line Profitability & Margin Performance
This notebook runs exploratory data analysis for the Nassau Candy Distributor project. It computes product- and division-level KPIs, Pareto concentration, cost diagnostics, and margin-risk scores, then surfaces charts and tables for the research report.

```python
# Imports
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import altair as alt
from src.metrics import compute_product_metrics, compute_pareto_products, compute_cost_diagnostics, compute_margin_risk

plt.style.use('seaborn')
```

```python
# Load sample data
DATA_PATH = 'data/raw/orders.csv'
df = pd.read_csv(DATA_PATH, parse_dates=['Order Date','Ship Date'])
print('Rows,cols:', df.shape)
df.head()
```

```python
# Compute metrics and tables
prod_metrics = compute_product_metrics(df)
pareto = compute_pareto_products(prod_metrics, revenue_col='Revenue', profit_col='Profit')
cost_diag = compute_cost_diagnostics(df)
risk = compute_margin_risk(df)

print('Products:', prod_metrics.shape[0])
prod_metrics.head()
```

```python
# Pareto chart (cumulative revenue & profit) using matplotlib
import matplotlib.ticker as mtick
fig, ax = plt.subplots(1,1,figsize=(8,4))
ax.plot(pareto['CumRevenuePct'].values, label='Cumulative Revenue')
ax.plot(pareto['CumProfitPct'].values, label='Cumulative Profit', color='orange')
ax.set_xlabel('Product Rank (by Revenue)')
ax.set_ylabel('Cumulative Share')
ax.yaxis.set_major_formatter(mtick.PercentFormatter(1.0))
ax.legend()
plt.tight_layout()
plt.show()

# Cost vs Revenue scatter (top products)
top = cost_diag.sort_values('Revenue', ascending=False).head(50)
fig2, ax2 = plt.subplots(1,1,figsize=(8,4))
sc = ax2.scatter(top['Cost'], top['Revenue'], c=top['Margin'], cmap='RdYlGn', s=80)
ax2.set_xlabel('Total Cost')
ax2.set_ylabel('Total Revenue')
plt.colorbar(sc, label='Margin')
plt.tight_layout()
plt.show()
```

## Key Findings (sample)
- Pareto: a small set of products account for the majority of revenue and profit (see Pareto chart).
- Cost Diagnostics: flagged high-cost / low-margin products require pricing or sourcing review.
- Margin Risk: products with high volatility and low profit contribution are higher risk and should be prioritized for review.
